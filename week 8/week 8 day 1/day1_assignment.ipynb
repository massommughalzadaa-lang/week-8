## Part 0 — Load the data (given)

Run this cell as-is. It loads the Wine dataset into a clean DataFrame for you.

```python
import pandas as pd
from sklearn.datasets import load_wine

# Load into a DataFrame
data = load_wine(as_frame=True)
wine = data.frame          # features + a 'target' column (0, 1, 2 = wine type)

wine.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

## Part 1 — Explore (2 marks)

Print the **shape** of the dataset, and use `value_counts()` on the `target` column to see how many wines of each type there are.

```python
# YOUR CODE HERE
wine.shape
print("shape of wine:", wine.shape)

# Split the data into features and target
print(wine["target"].value_counts())
```

## Part 2 — Split into X and y (1 mark)

- `X` = every column **except** `target`  *(hint: `wine.drop(columns="target")`)*
- `y` = the `target` column

```python
from sklearn.model_selection import train_test_split

X = wine.drop(columns="target")

y = wine["target"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

## Part 3 — Train / Test split (2 marks)

Hold out **20%** for testing. Use `random_state=42`.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.11, random_state=42
)

print("Training examples:", X_train.shape[0])
print("Testing examples: ", X_test.shape[0])
```

## Part 4 — Train a KNN model (1 mark)

Use `n_neighbors=5`. Fit it on the training data.

```python
from sklearn.neighbors import KNeighborsClassifier
# Create the model
model = KNeighborsClassifier(n_neighbors=5)

# Train it
model.fit(X_train, y_train)

print("Trained! ✅")
```

## Part 5 — Predict and score (2 marks)

Predict on the test set and print the accuracy as a percentage.

```python
from sklearn.metrics import accuracy_score

predictions = model.predict(X_test)
acc = accuracy_score(y_test, predictions)
print(f"Accuracy: {acc*100:.0f}%")

```
